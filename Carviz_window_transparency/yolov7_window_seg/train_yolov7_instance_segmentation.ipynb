### Data Format

**NOTE:** More about annotation formats: [Roboflow Computer Vision Annotation Formats](https://roboflow.com/formats).

1. Dataset directory structure

Dataset directory contains images and labels divided into three parts - train, test and validation sub-sets. In addition, there should be a `data.yaml` file in the dataset root directory.

```
HOME/
└── dataset-name/
    ├── train/
    │   ├── images/
    │   │   ├── image-0.jpg
    │   │   ├── image-1.jpg
    │   │   └── ...
    │   └── labels/
    │       ├── image-0.txt
    │       ├── image-1.txt
    │       └── ...
    ├── test/
    │   ├── images/
    │   │   └── ...
    │   └── labels/
    │       └── ...
    ├── valid/
    │   ├── images/
    │   │   └── ...
    │   └── labels/
    │       └── ...
    └── data.yaml
```

### Download dataset from Roboflow Universe

Roboflow API key: click on your profile in the upper right corner of the Roboflow app, then `Settings`. On the left, below `WORKSPACES` click in `Roboflow` -> `Roboflow API`. Copy the `Private API Key` for the code block below.

Note: your API key should be stored in a .env file, in the same directory as this jupyter notebook.

````python
!pip install roboflow==0.2.29 python-dotenv

from roboflow import Roboflow
from dotenv import load_dotenv
import os

load_dotenv()

rf = Roboflow(api_key=os.getenv('ROBOFLOW_API_KEY'))
project = rf.workspace("car-segmentation").project("car-o22yh")
version = project.version(1)
dataset = version.download("yolov7")

print("Dataset location:", dataset.location)
````

### Custom Training

````python
!nvidia-smi
````

````python
%cd {HOME}/yolov7/seg
!python3 segment/train.py --batch 16 \
 --epochs 10 \
 --data {dataset.location}/data.yaml \
 --weights $WEIGHTS_PATH \
 --device 0 \
 --name custom
````
